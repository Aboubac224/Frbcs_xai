```python
# Import banknote authentication loading function
from frbcs_xai.data.loader import load_uci_dataset

df, y = load_uci_dataset(267)

# The first three examples
df.head(3)
```

```python
import pandas as pd

# Number of missing values per variable
missing_count = df.isna().sum()

# Percentage of missing values
missing_pct = df.isna().mean() * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing (%)": missing_pct
})

print(missing_summary)
```

```python
# Number of missing values for target variable
missing_count_y = y.isna().sum()

# Percentage of missing values
missing_pct_y = y.isna().mean() * 100

missing_summary_y = pd.DataFrame({
    "Missing Count": missing_count_y,
    "Missing (%)": missing_pct_y
})

print(missing_summary_y)
```

```python
import numpy as np
import pandas as pd

# Get the class categories and the number of example per class
classes, counts = np.unique(y, return_counts=True)

class_distribution = pd.DataFrame({
    "Class": classes,
    "Count": counts,
    "Percentage (%)": counts / counts.sum() * 100
})

display(class_distribution)
```
